## 3. Import Data Generation Utilities

We use custom data generation functions that create realistic retail/CPG planning datasets.

````python
import numpy as np
import pandas as pd
import sys
import os
from sklearn.model_selection import train_test_split

# Add the scripts directory to the path so we can import util.py
# In Databricks, you may need to adjust this path based on your repo structure
repo_root = os.path.dirname(os.getcwd()) if 'notebooks' in os.getcwd() else os.getcwd()
scripts_path = os.path.join(repo_root, 'scripts')
if scripts_path not in sys.path:
    sys.path.insert(0, scripts_path)

# Import data generation functions
from util import (
    generate_supplier_delay_risk_data,
    generate_material_shortage_data,
    generate_labor_shortage_data,
    generate_price_elasticity_data,
    generate_promotion_lift_data,
    generate_supplier_lead_time_data,
    generate_transportation_lead_time_data,
    generate_yield_prediction_data,
    generate_scrap_anomaly_data,
    generate_aggregate_demand_forecast_data,
    generate_otif_risk_data
)

# Configuration for train/score split
TRAIN_RATIO = 0.8  # 80% for training, 20% for scoring

def save_train_score_split(df, table_name, stratify_col=None, random_state=42):
    """
    Split a DataFrame into train and score sets and save them as Delta tables.
    
    Args:
        df: pandas DataFrame to split
        table_name: base name for the table (will create table_name_train and table_name_score)
        stratify_col: column name to use for stratified split (for classification tasks)
        random_state: random seed for reproducibility
    """
    if stratify_col and stratify_col in df.columns:
        stratify = df[stratify_col]
    else:
        stratify = None
    
    df_train, df_score = train_test_split(
        df, 
        train_size=TRAIN_RATIO, 
        random_state=random_state,
        stratify=stratify
    )
    
    # Save train set
    spark.createDataFrame(df_train).write.mode("overwrite").saveAsTable(f"{table_name}_train")
    print(f"✓ Saved {len(df_train)} rows to {CATALOG}.{SCHEMA}.{table_name}_train")
    
    # Save score set
    spark.createDataFrame(df_score).write.mode("overwrite").saveAsTable(f"{table_name}_score")
    print(f"✓ Saved {len(df_score)} rows to {CATALOG}.{SCHEMA}.{table_name}_score")
    
    return df_train, df_score

print("Data generation utilities imported successfully!")
````

## 4. Demand Forecast Dataset (Time Series)

**Use Case:** Forecast product demand by category and region

This dataset contains monthly demand data across multiple product categories and regions, supporting demand forecasting and inventory planning processes.

````python
# Generate demand forecast data
df_demand_forecast = generate_aggregate_demand_forecast_data(n_series=50, n_months=36, seed=42)

print(f"Demand Forecast Dataset")
print(f"Shape: {df_demand_forecast.shape}")
print(f"Number of time series: {df_demand_forecast['series_id'].nunique()}")
print(f"Time range: {df_demand_forecast['date'].min()} to {df_demand_forecast['date'].max()}")
print(f"\nCategory distribution:")
print(df_demand_forecast['category'].value_counts())

display(df_demand_forecast.head(10))
````

````python
# Split by timestamp - each series is split into train (earlier) and score (later)
# This is the proper way to split time series data for forecasting

train_dfs = []
score_dfs = []

for series_id in df_demand_forecast['series_id'].unique():
    series_data = df_demand_forecast[df_demand_forecast['series_id'] == series_id].sort_values('date')
    n_points = len(series_data)
    split_idx = int(n_points * TRAIN_RATIO)
    
    train_dfs.append(series_data.iloc[:split_idx])
    score_dfs.append(series_data.iloc[split_idx:])

df_demand_train = pd.concat(train_dfs, ignore_index=True)
df_demand_score = pd.concat(score_dfs, ignore_index=True)

n_series = df_demand_forecast['series_id'].nunique()
train_cutoff = df_demand_train.groupby('series_id')['date'].max().min()
score_start = df_demand_score.groupby('series_id')['date'].min().max()

spark.createDataFrame(df_demand_train).write.mode("overwrite").saveAsTable("demand_forecast_train")
print(f"✓ Saved {len(df_demand_train)} rows ({n_series} series) to {CATALOG}.{SCHEMA}.demand_forecast_train")
print(f"  Train period: {df_demand_train['date'].min()} to {df_demand_train['date'].max()}")

spark.createDataFrame(df_demand_score).write.mode("overwrite").saveAsTable("demand_forecast_score")
print(f"✓ Saved {len(df_demand_score)} rows ({n_series} series) to {CATALOG}.{SCHEMA}.demand_forecast_score")
print(f"  Score period: {df_demand_score['date'].min()} to {df_demand_score['date'].max()}")
````

## Summary

All datasets have been prepared and saved as Delta tables in `tabular_fm.default`. Each dataset is split into **train** (80%) and **score** (20%) sets:

| Base Table | Task | Planning Process | Train | Score |
|------------|------|------------------|-------|-------|
| `demand_forecast`* | Time Series | Demand Planning | ~1,440 | ~360 |
| `price_elasticity` | Regression | Demand Planning | 2,400 | 600 |

| `transportation_lead_time` | Regression | Distribution Planning | 1,600 | 400 |
| `otif_risk` | Multi-class Classification | Distribution Planning | 1,600 | 400 |

**Split strategies:**
- **Time Series** (`demand_forecast`)*: Split by **timestamp** within each series. Earlier 80% of timestamps go to train, later 20% go to score. All series appear in both train and score sets.
- **Other datasets**: Random row-level split with stratification for classification tasks.

**Table naming convention:**
- `{table_name}_train` - Training data for model development
- `{table_name}_score` - Holdout data for scoring/inference
